# WDC Products — Development EDA

This notebook reads **train and validation only**. Test sets stay sealed for model evaluation. Run `python scripts/prepare_data.py` first.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

from productmatch.data.eda import development_summary

root = Path('../data/processed/wdc50_medium')
train = pd.read_parquet(root / 'train.parquet')
validation = pd.read_parquet(root / 'validation.parquet')
print(train.shape, validation.shape)
train.head(3)
```

```python
development_summary({'train': train, 'validation': validation})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, (name, frame) in zip(axes, [('train', train), ('validation', validation)], strict=True):
    frame['label'].value_counts().reindex([0, 1], fill_value=0).plot.bar(ax=ax)
    ax.set_title(f'{name} label balance')
    ax.set_xlabel('0 = nonmatch, 1 = match')
plt.tight_layout()
```

```python
train[['pair_id', 'label', 'title_left', 'title_right', 'is_hard_negative']].sample(
    n=min(10, len(train)), random_state=42
)
```
